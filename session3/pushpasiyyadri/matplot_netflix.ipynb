```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("netflix_titles.csv")
print(df)
```

```python
df.columns
```

```python
countries = df["country"].value_counts().head(10)

plt.bar(
    countries.index,
    countries.values
)

plt.title("Top 10 Countries")
plt.xlabel("Country")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.hist(df["release_year"])

plt.title("Release Year Distribution")
plt.xlabel("Release Year")
plt.ylabel("Count")
plt.grid()
plt.show()
```

```python
plt.boxplot(df["release_year"].dropna())
plt.show()
```

```python
df["type"].value_counts().plot.pie(autopct="%1.1f%%")
plt.title("Netflix content Type")
plt.grid()
plt.show()
```

```python
plt.bar(df["type"].value_counts().index,
        df["type"].value_counts().values)
plt.xlabel("type of values")
plt.ylabel("Type of content")
plt.title("Type of the contents and values")
plt.show()
```

```python
rating_count = df["rating"].value_counts()

plt.bar(
    rating_count.index,
    rating_count.values
)

plt.title("Netflix Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.show()
```

```python
year_count = (
    df["release_year"]
    .value_counts()
    .sort_index()
)

plt.plot(
    year_count.index,
    year_count.values
)

plt.title("Netflix Content by Year")
plt.xlabel("Year")
plt.ylabel("Number of Titles")
plt.grid()

plt.show()
```

```python
movies = df[df["type"] == "Movie"]



movie_year = movies["release_year"].value_counts().sort_index()


plt.plot(
    movie_year.index,
    movie_year.values,
    label="Movies"
)



plt.title("Movies  by Year")
plt.xlabel("Year")
plt.ylabel("Number of Titles")

plt.legend()

plt.show()
```

```python
movies = df[df["type"] == "Movie"]

shows = df[df["type"] == "TV Show"]

movie_year = movies["release_year"].value_counts().sort_index()
show_year = shows["release_year"].value_counts().sort_index()

plt.plot(
    movie_year.index,
    movie_year.values,
    label="Movies"
)

plt.plot(
    show_year.index,
    show_year.values,
    label="TV Shows"
)

plt.title("Movies vs TV Shows by Year")
plt.xlabel("Year")
plt.ylabel("Number of Titles")

plt.legend()

plt.show()
```
