```python
location_count = df["location"].value_counts().head(20)
plt.figure(figsize=(10, 6))
plt.plot(location_count.values,location_count.index,marker="o")
plt.xlabel("Number Of Hotels")
plt.ylabel("Location")
plt.title("Number of Hotels by Location")
plt.show()
```

```python
food_count = df["food_type"].value_counts().head(20)
plt.figure(figsize=(10, 6))
plt.bar(food_count.values,food_count.index)
plt.xlabel("Number of Hotels")
plt.ylabel("Food Type")
plt.title("Hotels by Food Type")
plt.show()
```

```python
food_count = df["food_type"].value_counts().head(5)
plt.figure(figsize=(8, 8))
plt.pie(food_count.values,labels=food_count.index,autopct="%1.1f%%")
plt.title("Top 5 Food Types")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df["rating_and_delivery_time"])
plt.xlabel("Rating")
plt.ylabel("Number of Hotels")
plt.title("Distribution of Hotel Ratings")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df["rating_and_delivery_time"].head(10),df["location"].head(10))
plt.xlabel("Rating and Delivery Time")
plt.ylabel("Location")
plt.title("Rating and Delivery Time by Location")
plt.show()
```
