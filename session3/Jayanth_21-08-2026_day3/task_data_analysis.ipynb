```python
for risk in df['diabetes_risk'].unique():
    df[df['diabetes_risk'] == risk]['age'].hist(
        bins=10,
        alpha=0.5,
        label=risk
    )

plt.legend()
plt.xlabel('Age')
plt.ylabel('Count')
plt.title('Age Distribution by Risk Factor')
plt.show()
```
