```python
from nltk.tokenize import word_tokenize
texts = [
    "I love Python.",
    "I love Generative AI.",
    "Python is useful for AI.",
    "I am learning tokenization."
]

for text in texts:
    tokens = word_tokenize(text)
    print("\nText:", text)
    print("Tokens:", tokens)
    print("Count:", len(tokens))
```

```python
from nltk.tokenize import word_tokenize
texts = [
    "happy",
    "unhappy",
    "happiness",
    "unhappiness",
    "programming",
    "programmer"
]
for text in texts:
    tokens = word_tokenize(text)
    print(text, "->", tokens)
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize

text = """
Python is easy to learn.
I am practicing GenAI.
Tokenization is my current topic.
"""

sentences = sent_tokenize(text)

print("Number of sentences:", len(sentences))

for sentence in sentences:
    tokens = word_tokenize(sentence)
    
    print("\nSentence:", sentence)
    print("Tokens:", tokens)
    print("Token count:", len(tokens))
```

```python
import string
from nltk.tokenize import word_tokenize

text = "Hello, I am learning Python!"

tokens = word_tokenize(text)

clean_tokens = [
    token for token in tokens
    if token not in string.punctuation
]

print("Original tokens:", tokens)
print("Without punctuation:", clean_tokens)
```

```python
import matplotlib.pyplot as plt
from nltk.tokenize import word_tokenize

text = "I love Python and Generative AI"

tokens = word_tokenize(text)

positions = list(range(len(tokens)))

plt.bar(positions, [1] * len(tokens))

plt.xticks(positions, tokens, rotation=45)
plt.xlabel("Token Position")
plt.ylabel("Token")
plt.title("Token Positions")

plt.show()
```

```python
from nltk.tokenize import word_tokenize
from collections import Counter
import matplotlib.pyplot as plt

text = """
Python is easy.
Python is powerful.
Python is useful for AI.
AI is useful for many applications.
"""

tokens = word_tokenize(text.lower())

counts = Counter(tokens)

print(counts)
```
