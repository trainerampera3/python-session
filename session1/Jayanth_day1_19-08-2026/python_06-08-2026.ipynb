# Hangman


```python
import random
```

```python
word_list = [
"awkward", "bagpipes", "crypt", "cycle", "dwarves",
"galaxy", "jukebox", "kayak", "oxygen", "pixel",
"puzzle", "rhythm", "sphinx", "wizard", "zombie"
]


chosen_word = random.choice(word_list)

placeholder = []

for i in range(len(chosen_word)):
    placeholder.append('_')

lives = 5
num = []
while(lives > 0 and  len(num) != len(chosen_word)):
    print(placeholder)
    
    guess = input('Guess the letter').lower()
    
    gr = False
    for i in range(len(chosen_word)):
        if i not in num and chosen_word[i] == guess:
            placeholder[i] = guess
            num.append(i)
            gr = True
            break;
    if not gr:
        lives -= 1
        gr = False

if(lives == 0):
    print('Your lives are gone! Better luck nex time')
else:
    print('Hurray! You have won the game')
        

    
```
