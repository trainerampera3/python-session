```python
def calculate_love_score(name1, name2):
    true = 0
    love = 0
    for i in name1:
        if i.lower() in 'true':
            true += 1
        if i.lower() in 'love':
            love += 1
    for i in name2:
        if i.lower() in 'true':
            true += 1
        if i.lower() in 'love':
            love += 1
    
    print("Love Score = "+str(true)+str(love))

calculate_love_score()
```

# Caesar Cipher

```python
alphabets = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm',
 'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z']

direction = input("Type 'encode' to encrypt , type 'decode to decrypt:\n'").lower()
text = input('Type your messsages:\n').lower()
shift = int(input('Type the shift number:\n'))

def encrypt(text, shift):
    msg = ""

    for i in text:
        if i in alphabets:
            k = alphabets.index(i)
            msg += alphabets[(k+shift)%25]
        else:
            msg += i

    print(msg)


def decrypt(text, shift):
    msg = ''
    for i in text:
        if i in alphabets:
            k = alphabets.index(i)
            k -= shift
            msg += alphabets[k % 25]
        else:
            msg += i

    print(msg)


def caesar(text, shift, encode_decode):
    if encode_decode == 'encode':
        encrypt(text, shift)
    elif encode_decode == 'decode':
        decrypt(text, shift)
    else:
        print('Wrong choice')

caesar(text, shift, direction)
        
```
